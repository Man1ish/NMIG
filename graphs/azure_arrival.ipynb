```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
path = "../dataset/functioninvocationtrace/AzureFunctionsInvocationTraceForTwoWeeksJan2021.txt"
```

```python
df = pd.read_csv(path)
```

```python
df['arrival_time'] = df['end_timestamp'] - df['duration']
df.sort_values(by=['arrival_time'], inplace=True, ascending=True)
df['arrival_time_fix'] = df['arrival_time'] - df['arrival_time'].iloc[0]
df.sort_values(by=['arrival_time_fix'], inplace=True, ascending=True)
df
```

```python
# Step 1: Round up the arrival time
df['arrival_time_ceil'] = np.ceil(df['arrival_time']).astype(int)

# Step 2: Group by 'func' and 'arrival_time_ceil' and count
grouped = df.groupby(['func', 'arrival_time_ceil']).size()

# Step 3: Find the maximum count per function
max_per_func = grouped.groupby('func').max()
max_per_func = max_per_func.sort_values(ascending=False)


max_per_func
```
