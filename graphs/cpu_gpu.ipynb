```python
import matplotlib.pyplot as plt
import pandas as pd

# Data
data = {
    "Models": ["alexnet", "resnet50", "efficientnet", "googlenet", "inception"],
    "CPU": [1.875, 1.693, 2.377, 1.57, 1.808],
    "GPU1": [2.634, 2.182, 2.621, 1.791, 2.236],
    "GPU2": [2.782, 2.758, 3.487, 2.363, 2.657]
}
df = pd.DataFrame(data)

# Plot
fig, ax = plt.subplots(figsize=(10, 6))
bar_width = 0.25
index = range(len(df["Models"]))

# Use IEEE-style colors: clean, professional, accessible
colors = ["#1f77b4", "#ff7f0e", "#2ca02c"]  # Blue, Orange, Green

bars1 = ax.bar([i - bar_width for i in index], df["CPU"], width=bar_width, label='CPU', color=colors[0])
bars2 = ax.bar(index, df["GPU1"], width=bar_width, label='GPU1', color=colors[1])
bars3 = ax.bar([i + bar_width for i in index], df["GPU2"], width=bar_width, label='GPU2', color=colors[2])

# Annotate bars with values
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 3),  # vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=9)

# Styling
ax.set_xlabel("Models", fontsize=12)
ax.set_ylabel("Execution Time (seconds)", fontsize=12)
ax.set_title("Inference Time Across CPU and GPUs", fontsize=14, fontweight='bold')
ax.set_xticks(index)
ax.set_xticklabels(df["Models"], fontsize=11)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
