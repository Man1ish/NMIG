```python
import matplotlib.pyplot as plt

# Data
tasks = ['Image', 'Video']
cpu_usage = [2.289, 26.325]
gpu_one = [2.512, 17.477]
gpu_two = [2.963, 25.194]

# Plotting
x = range(len(tasks))
width = 0.25

fig, ax = plt.subplots(figsize=(6, 4))
bars1 = ax.bar([p - width for p in x], cpu_usage, width, label='CPU')
bars2 = ax.bar(x, gpu_one, width, label='GPU-One')
bars3 = ax.bar([p + width for p in x], gpu_two, width, label='GPU-Two')

# IEEE-style formatting
ax.set_ylabel('Execution Time (s)', fontsize=10)
ax.set_xlabel('ML Functions', fontsize=10)
ax.set_xticks(x)
ax.set_xticklabels(tasks, fontsize=10)
ax.tick_params(axis='both', which='major', labelsize=9)
ax.legend(fontsize=9, loc='upper right')
ax.grid(axis='y', linestyle='--', linewidth=0.5, alpha=0.7)

# Remove top and right borders (IEEE-style minimalism)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
# plt.savefig("/mnt/data/cpu_gpu_ieee_style.jpg", dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
tasks = ['Image', 'Video']
cpu_usage = [2.289, 26.325]
gpu_one = [2.512, 17.477]
gpu_two = [2.963, 25.194]

x = np.arange(len(tasks))
width = 0.25

fig, ax = plt.subplots(figsize=(6, 3.5))

# Bar plots with hatch patterns for IEEE-friendly grayscale print
bars1 = ax.bar(x - width, cpu_usage, width, label='CPU', hatch='//', edgecolor='black', color='white')
bars2 = ax.bar(x, gpu_one, width, label='GPU (1x)', hatch='\\\\', edgecolor='black', color='white')
bars3 = ax.bar(x + width, gpu_two, width, label='GPU (2x)', hatch='xx', edgecolor='black', color='white')

# Add data labels to bars
ax.bar_label(bars1, fmt='%.2f', fontsize=9, padding=2)
ax.bar_label(bars2, fmt='%.2f', fontsize=9, padding=2)
ax.bar_label(bars3, fmt='%.2f', fontsize=9, padding=2)

# Axis labels and ticks
ax.set_ylabel('Execution Time (s)', fontsize=12)
ax.set_xlabel('ML Function Type', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(['Image Inference', 'Video Inference'], fontsize=9)
ax.tick_params(axis='y', labelsize=12)
ax.tick_params(axis='x', labelsize=12)

# IEEE-style: clean up
ax.legend(loc='upper center', bbox_to_anchor=(0.5, 1.2),
          fancybox=False, shadow=False, ncol=3, fontsize=12)
ax.grid(axis='y', linestyle='--', linewidth=0.5, alpha=0.7)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
# plt.savefig("cpu_gpu_alexnet.jpg", dpi=300, bbox_inches='tight')
plt.show()
```
