```python
# =============================================================================
# REUSE ANALYSIS -- Jupyter notebook version
# Paste each block below into its own cell. Run top to bottom.
# Dataframes stay in memory so you can inspect between cells.
# =============================================================================


# ---- CELL 1: imports + config ----------------------------------------------
import json
import numpy as np
import pandas as pd
 
CATEGORY = {
    "resnet50": 1, "googlenet": 1, "inception": 1,
    "alexnet": 2, "efficientnet": 2,
    "bert": 3, "distilgpt2": 3,
}
WINDOWS = [0.5, 1.0, 2.0]
 
 
# ---- CELL 2: helpers -------------------------------------------------------
def func_from_image(name):
    s = str(name)
    if ":" in s:
        s = s.split(":")[-1]
    if "-v" in s:
        s = s.split("-v")[0]
    return s.strip().lower()
 
def parse_epoch_ms(series):
    v = pd.to_numeric(series, errors="coerce").to_numpy(dtype=float)
    if np.nanmedian(v) > 1e11:
        v = v / 1000.0
    return v - np.nanmin(v)
 
 
# ---- CELL 3a: load docker log ----------------------------------------------
DOCKER_PATH = main_path+"docker_log.csv"
dock = pd.read_csv(DOCKER_PATH, header=None,
                   names=["timestamp", "container", "image", "status"])
dock["t"] = parse_epoch_ms(dock["timestamp"])
dock["func"] = dock["image"].map(func_from_image)
dock["status"] = dock["status"].astype(str).str.strip().str.lower()
print("docker funcs:", sorted(dock["func"].unique()))
 
 
# ---- CELL 3b: load the hash->func map (the table: id, func, path, ...) ------
MAP_PATH = main_folder+"functions.csv"
fmap = pd.read_csv(MAP_PATH)
fmap.columns = [c.strip() for c in fmap.columns]
hash2func = dict(zip(fmap["id"].astype(str).str.strip(),
                     fmap["func"].astype(str).str.strip().str.lower()))
print(f"map has {len(hash2func)} hash->func entries:")
for h, f in hash2func.items():
    print(f"  {h[:16]}...  ->  {f}")
 
 
# ---- CELL 3c: load activation log + map hash -> function -------------------
ACTS_PATH =  main_path+"results_updated.csv"
acts = pd.read_csv(ACTS_PATH)
acts.columns = [c.strip() for c in acts.columns]
 
# the column holding the hash. From your file this is 'func'. change if needed.
HASH_COL = "func"
TIME_COL = "timestamp"
 
acts["t"] = parse_epoch_ms(acts[TIME_COL])
acts["func"] = acts[HASH_COL].astype(str).str.strip().map(hash2func)
 
unmapped = acts["func"].isna().sum()
print(f"\nmapped {len(acts) - unmapped}/{len(acts)} activations; {unmapped} unmapped")
if unmapped:
    ex = acts.loc[acts["func"].isna(), HASH_COL].iloc[0]
    print("[WARN] some hashes not in the map -> WRONG map version for this run.")
    print("       example unmapped hash:", ex)
    print("       (v1.0 vs v1.2 runs have different hashes -- use the matching map)")
acts["func"] = acts["func"].fillna("unknown")
print("funcs seen:", sorted(acts["func"].unique()))
 
bad = [f for f in acts["func"].unique() if f not in CATEGORY and f != "unknown"]
if not unmapped and not bad:
    print("[OK] every activation mapped to one of the 7 known functions.")
 
 
# ---- CELL 4: container creations -------------------------------------------
creations = dock.sort_values("t").drop_duplicates("container")[["t", "func", "container"]]
creations = creations.reset_index(drop=True)
print("total container creations:", len(creations))
print(creations["func"].value_counts())
 
 
# ---- CELL 5: reuse computation ---------------------------------------------
def compute_reuse(acts, dock, creations, window):
    cre_by_func = {f: np.sort(creations.loc[creations["func"] == f, "t"].to_numpy())
                   for f in creations["func"].unique()}
    alive_events = (dock[dock["status"].isin({"running", "paused"})]
                    .groupby("func")["t"].min().to_dict())
 
    ta = acts["t"].to_numpy()
    fa = acts["func"].to_numpy()
    total = len(acts)
    served_by_new = cross_cat = same_func = 0
 
    for i in range(total):
        t_i, f_i = ta[i], fa[i]
        cat_i = CATEGORY.get(f_i)
        arr = cre_by_func.get(f_i, np.empty(0))
        new_same = False
        if arr.size:
            lo = np.searchsorted(arr, t_i - window, "left")
            hi = np.searchsorted(arr, t_i + window, "right")
            new_same = hi > lo
        if new_same:
            served_by_new += 1
            continue
        mate = any(f != f_i and CATEGORY.get(f) == cat_i and start <= t_i
                   for f, start in alive_events.items())
        if mate:
            cross_cat += 1
        elif alive_events.get(f_i, np.inf) <= t_i:
            same_func += 1
 
    reuse = total - served_by_new
    return {
        "window_s": window,
        "total_activations": int(total),
        "served_by_new_container": int(served_by_new),
        "reuse_count": int(reuse),
        "reuse_rate": round(reuse / total, 4) if total else 0.0,
        "cross_category_reuse": int(cross_cat),
        "cross_category_reuse_rate": round(cross_cat / total, 4) if total else 0.0,
        "same_function_reuse": int(same_func),
        "total_creations": int(len(creations)),
        "creations_per_100_act": round(100 * len(creations) / total, 2) if total else 0.0,
    }
 
results = [compute_reuse(acts, dock, creations, w) for w in WINDOWS]
for r in results:
    print(f"window={r['window_s']}s  reuse_rate={r['reuse_rate']:.1%}  "
          f"cross_cat={r['cross_category_reuse']} ({r['cross_category_reuse_rate']:.1%})  "
          f"creations/100act={r['creations_per_100_act']}")
pd.DataFrame(results)
 
 
# ---- CELL 5b: per-function creation breakdown (the real reuse story) -------
# How many times did each function trigger a NEW container vs total requests.
per_func = []
act_counts = acts["func"].value_counts().to_dict()
cre_counts = creations["func"].value_counts().to_dict()
for f in sorted(set(act_counts) | set(cre_counts)):
    n_act = act_counts.get(f, 0)
    n_cre = cre_counts.get(f, 0)
    per_func.append({
        "func": f,
        "category": CATEGORY.get(f),
        "activations": n_act,
        "containers_created": n_cre,
        "creations_per_act": round(n_cre / n_act, 3) if n_act else None,
    })
per_func_df = pd.DataFrame(per_func).sort_values("category")
print(per_func_df.to_string(index=False))


# # ---- CELL 7: save (optional) -----------------------------------------------
# LABEL = "openwhisk"   # change per system: openwhisk / nmig / histogram / pagurus
# with open(f"reuse_{LABEL}.json", "w") as f:
#     json.dump(results, f, indent=2)
# print("saved", f"reuse_{LABEL}.json")
```
